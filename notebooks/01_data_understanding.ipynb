```python
import os
import pandas as pd
import numpy as np

# Define relative path to the raw dataset
DATA_PATH = os.path.join("..", "data", "raw", "credit_risk_dataset.csv")

def inspect_raw_data(file_path: str):
    """
    Loads and performs initial audit on raw loan dataset.
    """
    if not os.path.exists(file_path):
        # Fallback check if script is run from project root instead of notebooks/
        file_path = os.path.join("data", "raw", "credit_risk_dataset.csv")

    print(f"--- Loading Data from: {file_path} ---")
    df = pd.read_csv(file_path)

    print("\n1. Dataset Shape:")
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")

    print("\n2. Data Types and Non-Null Counts:")
    print(df.info())

    print("\n3. Missing Value Audit:")
    missing = df.isnull().sum()
    missing_pct = (missing / len(df)) * 100
    missing_df = pd.DataFrame({'Missing Counts': missing, 'Percentage (%)': missing_pct.round(2)})
    print(missing_df[missing_df['Missing Counts'] > 0])

    print("\n4. Target Variable Class Distribution ('loan_status'):")
    target_counts = df['loan_status'].value_counts()
    target_pct = df['loan_status'].value_counts(normalize=True) * 100
    target_df = pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_pct.round(2)})
    print(target_df)

    print("\n5. Numerical Summary Statistics:")
    print(df.describe().T.round(2))

    print("\n6. First 5 Rows:")
    print(df.head())

    return df

if __name__ == "__main__":
    df = inspect_raw_data(DATA_PATH)
```
